# EDA — Opyna
Análisis exploratorio de la muestra de Amazon Reviews 2023 (Electronics).

> Requiere ejecutar antes: `python -m opyna.data.download --category Electronics`

```python
import sys; sys.path.insert(0, 'src')
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style='whitegrid')
df = pd.read_parquet('data/processed/reviews_electronics.parquet')
print('Shape:', df.shape)
print('\nNulos por columna:'); print(df.isna().sum())
df['n_words'] = df['text'].str.split().str.len()
print('\nLongitud (palabras):'); print(df['n_words'].describe().round(1))
print('\n% compra verificada:', round(df['verified_purchase'].mean()*100,1))
print('\nTop 5 productos:'); print(df['parent_asin'].value_counts().head(5))
```

```python
fig, ax = plt.subplots(2,2, figsize=(13,9))
sns.countplot(x='rating', hue='rating', data=df, palette='viridis', legend=False, ax=ax[0,0])
ax[0,0].set_title('Distribución de ratings (balanceada por diseño)')
sns.histplot(df['n_words'].clip(upper=300), bins=40, color='teal', ax=ax[0,1])
ax[0,1].set_title('Longitud de reseña (palabras, recorte a 300)')
sns.boxplot(x='rating', y='n_words', hue='rating', data=df[df.n_words<300], palette='viridis', legend=False, ax=ax[1,0])
ax[1,0].set_title('Longitud vs rating')
vc = df['verified_purchase'].value_counts()
ax[1,1].pie(vc.values, labels=['Verificada' if i else 'No verif.' for i in vc.index], autopct='%1.0f%%')
ax[1,1].set_title('Compra verificada')
plt.tight_layout(); plt.show()
```
